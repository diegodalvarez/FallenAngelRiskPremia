```python
path = os.path.join(raw_path, "ETFData.parquet")
(pd
    .read_parquet(path = path, engine = "pyarrow")
    .security
    .drop_duplicates())
```

```python
cred_path   = os.path.join(raw_path, "PX.parquet")
df_cred_rtn = (pd
    .read_parquet(path = cred_path, engine = "pyarrow")
    .loc[lambda x: x.Price == "Adj Close"]
    .pivot(index = "date", columns = "ticker", values = "value")
    .pct_change()
    .reset_index()
    .melt(id_vars = "date")
    .dropna())
```
